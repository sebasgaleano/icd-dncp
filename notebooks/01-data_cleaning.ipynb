**Dataset:** Datos abiertos - DNCP Paraguay (2021-2022)
**Fuente** https://contrataciones.gov.py/datos

## 1.1 - Objetivo
Construir un dataset a partir de los datos consolidados de los archivos csv crudos de adjudicaciones y contratos de los años 2021 y 2022, listo para el análisis exploratio.

## 1.2 - Estructura de los datos.
| datos
|-adjudicaciones
|-- 2021.csv # adjudicaciones realizadas en el año 2021
|-- 2022.csv # adjudicaciones realizadas en el año 2022
|- contratos
|-- 2021.csv # contratos realizados en el año 2021
|-- 2022.csv # contratos realizados en el año 2022

```python
# se importan pandas y numpy para el manejo de datos
# se definen las rutas de los archivos csv que contienen los datos de adjudicaciones y contratos para los años 2021 y 2022

import pandas as pd
import numpy as np

pd.options.display.float_format = '{:_.2f}'.format
 
adj_2021 = "../datos/adjudicaciones/2021.csv"
adj_2022 = "../datos/adjudicaciones/2022.csv"
cont_2021 = "../datos/contratos/2021.csv"
cont_2022 = "../datos/contratos/2022.csv"


```

```python
def csv_to_df(file_path):
    """
    Lee un archivo CSV y devuelve un DataFrame.
    Args:
        file_path (str): La ruta al archivo CSV.    
    Returns:
        pd.DataFrame: Un DataFrame con los datos del archivo CSV.   

    """
    return pd.read_csv(
        file_path,
        sep=",",
        encoding="utf-8",
        decimal=".",
        thousands=","
    )

df_adj_2021 = csv_to_df(adj_2021)
df_adj_2022 = csv_to_df(adj_2022)
df_cont_2021 = csv_to_df(cont_2021)
df_cont_2022 = csv_to_df(cont_2022)

data = {
    'df_adj_2021': df_adj_2021,
    'df_adj_2022': df_adj_2022,
    'df_cont_2021': df_cont_2021,
    'df_cont_2022': df_cont_2022,
}
for name, df in data.items():
    print(f'datos de csv "{name}": | {df.shape[0]} filas | {df.shape[1]} columnas')


```

```python
for name, df in data.items():
    print(f"{name}: {df.columns.tolist()}" )
```

```python
for name, df in data.items():
    print(f'columnas del dataframe "{name}": {df.head(3)}')
```

```python
df_adj = pd.concat([df_adj_2021, df_adj_2022], ignore_index=True)
df_cont = pd.concat([df_cont_2021, df_cont_2022], ignore_index=True)

df = df_adj.merge(df_cont[["id_llamado", "ruc", "razon_social", 
                             "monto_adjudicado", "fecha_firma_contrato", 
                             "vigencia_contrato"]], 
                  on="id_llamado", how="left")
print(f"DataFrame final: {df.shape[0]} filas | {df.shape[1]} columnas")
```

```python
df_adj.shape, df_cont.shape
```

```python
# Cuántos id_llamado únicos en cada tabla
print("adj únicos:", df_adj["id_llamado"].nunique())
print("cont únicos:", df_cont["id_llamado"].nunique())

# Cuántos de adj no matchearon con cont
sin_match = df_adj[~df_adj["id_llamado"].isin(df_cont["id_llamado"])]
print("adj sin match en cont:", len(sin_match))
```

```python
print(df.isnull().sum().sort_values(ascending=False))
print("\n% nulos:")
print((df.isnull().sum() / len(df) * 100).sort_values(ascending=False).round(2))
```

```python
cols_drop = ["organismo_financiador", "organismo_financiador_id", 
             "vigencia_contrato", "observaciones"]
df = df.drop(columns=cols_drop)

df["fecha_firma_contrato"] = pd.to_datetime(df["fecha_firma_contrato"], errors="coerce")
df["fecha_publicacion"] = pd.to_datetime(df["fecha_publicacion"], errors="coerce")

print(df.shape)
```

```python
# Montos en 0
print("monto_total_adjudicado == 0:", (df["monto_total_adjudicado"] == 0).sum())
print("monto_adjudicado == 0:", (df["monto_adjudicado"] == 0).sum())

# Duplicados por id_llamado
print("\nDuplicados por id_llamado:", df.duplicated(subset="id_llamado").sum())
print("Duplicados por id (fila exacta):", df.duplicated().sum())
```

```python
df[df["monto_total_adjudicado"] == 0][["id_llamado", "nombre_licitacion", 
                                        "tipo_procedimiento", "convocante"]].head(10)
```

```python
df[df["id_llamado"] == df[df.duplicated(subset="id_llamado")]["id_llamado"].iloc[0]][
    ["id_llamado", "razon_social", "monto_adjudicado"]
]
```

```python
# Cuántos tienen tipo_procedimiento específico
df[df["monto_total_adjudicado"] == 0]["tipo_procedimiento"].value_counts()
```

```python
df = df.drop_duplicates()
df = df[df["monto_total_adjudicado"] > 0]
print(df.shape)
```

```python
oferentes_por_llamado = (
    df.groupby("id_llamado")["ruc"]
    .nunique()
    .reset_index()
    .rename(columns={"ruc": "num_oferentes"})
)

df = df.merge(oferentes_por_llamado, on="id_llamado", how="left")

print(df["num_oferentes"].describe())
print(df["num_oferentes"].value_counts().head())
```

```python

df["competencia"] = (df["num_oferentes"] > 1).astype(int)

# Los 342 con num_oferentes == 0 quedan competencia = 0
# los excluiremos en Etapa 3 explícitamente

print(df["competencia"].value_counts())
print(df["competencia"].value_counts(normalize=True).round(3))
```

```python
df.to_csv("../datos/df_clean.csv", index=False)
```
